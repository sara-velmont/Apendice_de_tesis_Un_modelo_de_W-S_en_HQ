```python
# Función para seleccionar enlaces de E con probabilidad p
def enlaces_prob(E, p):
    for enlace in E:
        if np.random.rand() < p:
            yield enlace

# Función para convertir un conjunto como {'A','B'} a 'AB'
def etiqueta(nodo):
    return ''.join(sorted(nodo))

# Construcción de la red aleatoria con etiquetas legibles
def hg_aleatorio(E, P, p):
    lista_enlaces = list(enlaces_prob(E, p))

    G = nx.Graph()
    nodos = [frozenset(n) for n in P]
    G.add_nodes_from(nodos)
    G.add_edges_from([(frozenset(x), frozenset(y)) for x, y in lista_enlaces])

    # Crear etiquetas legibles para los nodos
    etiquetas = {n: etiqueta(n) for n in G.nodes}

    # Dibujar el grafo
    plt.figure(figsize=(10, 8))
    pos = nx.kamada_kawai_layout(G)
    #pos = nx.spring_layout(G, seed=42)
    nx.draw(G, pos, with_labels=True, labels=etiquetas, node_size=800, node_color='green', font_size=10)
    plt.title("Red aleatoria H(E,p)")
    plt.show()

    return G
```
