#Entropía de Shannon

```python
for i, adj_matrix in enumerate(matrices):
    # Obtener autovalores
    eigvals, _ = np.linalg.eigh(adj_matrix)
    valores, multiplicidades = np.unique(np.round(eigvals, decimals=8), return_counts=True)
    N = np.sum(multiplicidades)

    # Dominio de omega
    omega = np.linspace(0, 10, 1000)
    delta_omega = omega[1] - omega[0]  # paso de integración

    # Construir f(omega)
    f_omega = np.zeros_like(omega, dtype=complex)
    for val, mult in zip(valores, multiplicidades):
        coef = np.sqrt(mult) / np.sqrt(N)
        f_omega += coef * np.exp(1j * val * omega)

    # Calcular |f(omega)|^2
    power_spectrum = np.abs(f_omega)**2

    # Área bajo la curva A
    Ar = np.trapezoid(power_spectrum, omega)  # integral numérica de |f(ω)|^2

    # Evitar problemas con log(0)
    epsilon = 1e-12
    normalized_density = power_spectrum / Ar + epsilon

    # Cálculo de la entropía
    integrand = (power_spectrum / 2) * np.log(normalized_density)
    entropy = -np.trapezoid(integrand, omega)

    # Imprimir resultado
    print(f"Matriz n={i+3}: entropía de Shannon ≈ {entropy:.6f}")
```

#Distancia de Kullback-Leibler

```python
omega = np.linspace(0, 10, 1000)
delta_omega = omega[1] - omega[0]
epsilon = 1e-12  # para evitar log(0)

def calcular_p_omega(adj_matrix):
    eigvals, _ = np.linalg.eigh(adj_matrix)
    valores, multiplicidades = np.unique(np.round(eigvals, decimals=8), return_counts=True)
    N = np.sum(multiplicidades)

    f_omega = np.zeros_like(omega, dtype=complex)
    for val, mult in zip(valores, multiplicidades):
        coef = np.sqrt(mult) / np.sqrt(N)
        f_omega += coef * np.exp(1j * val * omega)

    power_spectrum = np.abs(f_omega)**2
    Ar = np.trapz(power_spectrum, omega)  # integral numérica

    p_omega = power_spectrum / Ar + epsilon  # densidad normalizada y evitar log(0)
    return p_omega

# Precomputar todas las densidades p_n
densidades = [calcular_p_omega(A) for A in matrices]

# Función para calcular KL discretamente con integración por trapecios
def distancia_KL(p, q):
    # p y q deben ser arrays con misma dimensión, normalizados y sin ceros
    return np.trapz(p * np.log(p / q), omega)

# Calcular matriz de distancias KL
n_matrices = len(matrices)
kl_matrix = np.zeros((n_matrices, n_matrices))

for i in range(n_matrices):
    for j in range(n_matrices):
        if i != j:
            kl_matrix[i, j] = distancia_KL(densidades[i], densidades[j])
        else:
            kl_matrix[i, j] = 0.0  # KL entre igual es cero

# Mostrar resultados por pares:
for i in range(n_matrices):
    for j in range(n_matrices):
        if i != j:
            print(f"KL(p_{i+3} || p_{j+3}) = {kl_matrix[i,j]:.6f}")
```

Entropía de Shannon

```python
for i, adj_matrix in enumerate(matrices):
    # Obtener autovalores
    eigvals, _ = np.linalg.eigh(adj_matrix)
    valores, multiplicidades = np.unique(np.round(eigvals, decimals=8), return_counts=True)
    N = np.sum(multiplicidades)

    # Dominio de omega
    omega = np.linspace(0, 10, 1000)
    delta_omega = omega[1] - omega[0]  # paso de integración

    # Construir f(omega)
    f_omega = np.zeros_like(omega, dtype=complex)
    for val, mult in zip(valores, multiplicidades):
        coef = np.sqrt(mult) / np.sqrt(N)
        f_omega += coef * np.exp(1j * val * omega)

    # Calcular |f(omega)|^2
    power_spectrum = np.abs(f_omega)**2

    # Área bajo la curva A
    Ar = np.trapezoid(power_spectrum, omega)  # integral numérica de |f(ω)|^2

    # Evitar problemas con log(0)
    epsilon = 1e-12
    normalized_density = power_spectrum / Ar + epsilon

    # Cálculo de la entropía
    integrand = (power_spectrum / 2) * np.log(normalized_density)
    entropy = -np.trapezoid(integrand, omega)

    # Imprimir resultado
    print(f"Matriz A={i}: entropía de Shannon ≈ {entropy:.6f}")
```

Divergendia de Kullback-Leibler con respecto al HQC de $n=6$

```python
matrices = [A1, AR1, AR2, AR3, AR4, AR5, AR6, AR7, AR8, AR9, AR10, AR11]
omega = np.linspace(0, 10, 1000)
delta_omega = omega[1] - omega[0]
epsilon = 1e-12  # para evitar log(0)

def calcular_p_omega(adj_matrix):
    eigvals, _ = np.linalg.eigh(adj_matrix)
    valores, multiplicidades = np.unique(np.round(eigvals, decimals=8), return_counts=True)
    N = np.sum(multiplicidades)

    f_omega = np.zeros_like(omega, dtype=complex)
    for val, mult in zip(valores, multiplicidades):
        coef = np.sqrt(mult) / np.sqrt(N)
        f_omega += coef * np.exp(1j * val * omega)

    power_spectrum = np.abs(f_omega)**2
    Ar = np.trapz(power_spectrum, omega)  # integral numérica

    p_omega = power_spectrum / Ar + epsilon  # densidad normalizada y evitar log(0)
    return p_omega

# Precomputar todas las densidades p_n
densidades = [calcular_p_omega(A) for A in matrices]

# Función para calcular KL discretamente con integración por trapecios
def distancia_KL(p, q):
    # p y q deben ser arrays con misma dimensión, normalizados y sin ceros
    return np.trapz(p * np.log(p / q), omega)

# Calcular matriz de distancias KL
n_matrices = len(matrices)
kl_matrix = np.zeros((n_matrices, n_matrices))

for i in range(n_matrices):
    for j in range(n_matrices):
        if i != j:
            kl_matrix[i, j] = distancia_KL(densidades[i], densidades[j])
        else:
            kl_matrix[i, j] = 0.0  # KL entre igual es cero

# Mostrar resultados por pares:
for i in range(n_matrices):
    for j in range(n_matrices):
        if i != j:
            print(f"KL(p_{i+3} || p_{j+3}) = {kl_matrix[i,j]:.6f}")
```

Análisis espectral de HQ aleatorios (E-R) de $n=6$.

```python

# Selecciona cada enlace con probabilidad p
def enlaces_prob(E1, p):
    for enlace in E1:
        if np.random.rand() < p:
            yield enlace

```

```python
P1_frozen = [frozenset(n) for n in P1]
```

```python
def hg_aleatorio(E1, P1_frozen, p):

    G = nx.Graph()
    G.add_nodes_from(P1_frozen)

    G.add_edges_from(
        (frozenset(x), frozenset(y))
        for x, y in enlaces_prob(E1, p)
    )

    return G
```

```python

p_values = np.logspace(-3, 0, 10)

for p in p_values:

    for _ in range(100):

        G = hg_aleatorio(E1, P1_frozen, p)

        A = nx.to_numpy_array(
            G,
            nodelist=P1_frozen
        )

        eigvals = np.linalg.eigvalsh(A)
```

```python


# Dominio de Fourier
omega = np.linspace(0, 2, 1000)

# Número de realizaciones
n_ensamble = 100

# Para guardar las funciones promedio
f_promedios = {}


for p in p_values:

    f_media = np.zeros(len(omega), dtype=complex)

    for _ in range(n_ensamble):

        # Generar red
        G = hg_aleatorio(E1, P1_frozen, p)

        # Matriz de adyacencia
        A = nx.to_numpy_array(G, nodelist=P1_frozen)

        # Espectro
        eigvals = np.linalg.eigvalsh(A)

        valores, multiplicidades = np.unique(
            np.round(eigvals,8),
            return_counts=True
        )

        N = multiplicidades.sum()

        # Función espectral
        f = np.zeros(len(omega), dtype=complex)

        for val, mult in zip(valores, multiplicidades):
            coef = np.sqrt(mult)/np.sqrt(N)
            f += coef*np.exp(1j*val*omega)

        f_media += f

    f_media /= n_ensamble

    f_promedios[p] = f_media
```

```python
plt.figure(figsize=(10,6))

for p in p_values:
    plt.plot(
        omega,
        np.abs(f_promedios[p]),
        label=f"p={p:.3e}"
    )

plt.xlabel(r'$\omega$')
plt.ylabel(r'$|\,\overline{f}(\omega)\,|$')
plt.title('Función espectral promedio')
plt.legend(fontsize=8)
plt.grid(alpha=0.3)

plt.show()
```

```python


espectros_promedio = {}

for p in p_values:

    espectros = []

    for _ in range(100):

        G = hg_aleatorio(E1, P1_frozen, p)

        A = nx.to_numpy_array(G, nodelist=P1_frozen)

        eigvals = np.linalg.eigvalsh(A)

        # Ordenar los eigenvalores
        espectros.append(np.sort(eigvals))

    espectros = np.array(espectros)

    # Eigenvalores promedio
    espectros_promedio[p] = espectros.mean(axis=0)
```
